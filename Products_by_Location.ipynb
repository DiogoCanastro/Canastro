## Load Data from Excel Sheet

### Subtask:
Load the data from the 'Location Product' sheet (sheet number 3) of the '/content/Location Product MDs IBP 1.xlsx' file into a pandas DataFrame.


**Reasoning**:
To load the data from the specified Excel sheet into a pandas DataFrame and display its head and columns, I will first import the pandas library, then use `pd.read_excel` to load the data, and finally `df.head()` and `df.columns` to verify the load.



```python
import pandas as pd

# Load the data from the 'Location Product' sheet (sheet number 3)
df_location_product = pd.read_excel('/content/Location Product MDs IBP 1.xlsx', sheet_name='Location Product')

# Display the first few rows of the DataFrame
print("First 5 rows of the DataFrame:")
print(df_location_product.head())

# Display the columns of the DataFrame
print("\nColumns in the DataFrame:")
print(df_location_product.columns)
```

**Reasoning**:
The previous output shows that the first row of the DataFrame contains the true headers ('Location ID*', 'Product ID*'), and the actual data begins from the second row. Also, the column names are currently `LOCID` and `PRDID`. To correctly extract `Location ID` and `Product ID`, I need to reload the Excel sheet by skipping the first row, select only the relevant columns, drop any rows with missing values in these columns, and then rename them to 'Location ID' and 'Product ID' as per the task.



```python
import pandas as pd

# Load the data again, skipping the first row to use the second row as headers
df_location_product_clean = pd.read_excel(
    '/content/Location Product MDs IBP 1.xlsx',
    sheet_name='Location Product',
    header=1 # Set header to 1 (0-indexed) to use the second row as column names
)

# Select only the 'Location ID*' and 'Product ID*' columns
df_location_product_clean = df_location_product_clean[['Location ID*', 'Product ID*']]

# Drop rows where either 'Location ID*' or 'Product ID*' is missing
df_location_product_clean.dropna(subset=['Location ID*', 'Product ID*'], inplace=True)

# Rename the columns for clarity as per the task description
df_location_product_clean.rename(columns={'Location ID*': 'Location ID', 'Product ID*': 'Product ID'}, inplace=True)

# Display the first few rows of the cleaned DataFrame
print("First 5 rows of the cleaned DataFrame:")
print(df_location_product_clean.head())

# Display the shape of the cleaned DataFrame to see the number of rows and columns
print("\nShape of the cleaned DataFrame:")
print(df_location_product_clean.shape)
```

**Reasoning**:
Now that the data containing 'Location ID' and 'Product ID' has been loaded and cleaned, the next step is to create a function that takes a `Location ID` as input and returns a list of associated `Product ID`s, as specified in the main task. This function will use the `df_location_product_clean` DataFrame to filter products by location.



```python
def get_products_by_location(location_id):
    """
    Takes a Location ID as input and returns a list of associated Product ID's.
    """
    # Filter the DataFrame for the given location_id
    products_at_location = df_location_product_clean[df_location_product_clean['Location ID'] == location_id]

    # Extract the 'Product ID' column and convert it to a list
    product_ids = products_at_location['Product ID'].tolist()

    return product_ids

# Demonstrate the function with an example
example_location_id = '0100' # Using an example Location ID found in the data
associated_products = get_products_by_location(example_location_id)

print(f"Products associated with Location ID '{example_location_id}':")
print(associated_products)

# Display the number of products found for the example location
print(f"\nNumber of products found for Location ID '{example_location_id}': {len(associated_products)}")
```

```python
example_location_id = '2661' # Using an example Location ID found in the data
associated_products = get_products_by_location(example_location_id)

print(f"Products associated with Location ID '{example_location_id}':")
print(associated_products)
```

```python
import pandas as pd

# Convert the list of associated products to a DataFrame
df_associated_products = pd.DataFrame(associated_products, columns=['Product ID'])

# Define the output file path
output_file_path = 'associated_products.xlsx'

# Export the DataFrame to an Excel file
df_associated_products.to_excel(output_file_path, index=False)

print(f"Successfully exported {len(associated_products)} product IDs to '{output_file_path}'")
```
